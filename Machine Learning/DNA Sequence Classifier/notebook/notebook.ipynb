```python
TARGET='Class_Label'

y=LabelEncoder().fit_transform(df[TARGET])

kmer=CountVectorizer(analyzer='char', ngram_range=(3, 4), lowercase=False)
X_kmer=kmer.fit_transform(df["Sequence"])
X_kmer=X_kmer.multiply(1/X_kmer.sum(axis=1)).tocsr()

num_cols=['GC_Content', 'Sequence_Length', 'kmer_3_freq', 'Mutation_Flag']
X=hstack([X_kmer, csr_matrix(df[num_cols].values)]).tocsr()

cv=StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

models={
    "Baseline (Majority)": DummyClassifier(strategy='most_frequent'),
    "LogisticRegression":LogisticRegression(max_iter=1000, random_state=42),
    "XGBoost":XGBClassifier(use_label_encoder=False, eval_metric='logloss', random_state=42)
}

for name, m in models.items():
    print(f"{name:20s} {cross_val_score(m, X, y, cv=cv).mean():.3f}")
```

```python
import joblib
from pathlib import Path

# Keep the label encoder so the API can turn 0,1,2,3 back into class names
le = LabelEncoder()
y = le.fit_transform(df[TARGET])

# The API can only use features it can calculate from the user's input,
# so kmer_3_freq is left out here
api_cols = ["GC_Content", "Sequence_Length", "Mutation_Flag"]
X_api = hstack([X_kmer, csr_matrix(df[api_cols].values)]).tocsr()

final_model = XGBClassifier(n_estimators=300, max_depth=5, learning_rate=0.1)
cv_acc = cross_val_score(final_model, X_api, y, cv=cv).mean()
print(f"API model CV accuracy: {cv_acc:.3f}")

final_model.fit(X_api, y)  # train on all data

out = Path("../model.joblib")   # saves next to main.py
joblib.dump({
    "vectorizer": kmer,
    "model": final_model,
    "classes": list(le.classes_),
    "cv_accuracy": float(cv_acc),
}, out)
print("Saved to", out.resolve())
```
